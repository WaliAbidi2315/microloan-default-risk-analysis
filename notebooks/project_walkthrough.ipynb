## 4. Limitations

The generator encodes the relationship between predictors and target. A high score therefore cannot establish generalisation to real borrowers. The 180-default dataset and 36-positive test set also make estimates uncertain. Real deployment would require external/temporal validation, calibration, fairness assessment and a cost-based threshold policy.
